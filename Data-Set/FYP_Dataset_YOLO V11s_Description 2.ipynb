### Project Overview

This project develops a computer vision based mobile application for verifying Pakistani banknotes by detecting denomination and important visible security features. The application is an academic FYP prototype, not a certified currency authentication system.

The proposed pipeline uses YOLO11s for object detection and can use MobileNetV3 as a complementary image classification model. The final application combines model predictions with rule based feature verification.


# 1. Dataset Description

The project uses Pakistani banknote images containing different denominations and annotated security features. The images are prepared for object detection using bounding box annotations.

The main denominations are:

- Rs. 10
- Rs. 20
- Rs. 50
- Rs. 100
- Rs. 500
- Rs. 1000
- Rs. 5000

The dataset includes photographs captured under different conditions because the final system is intended to process mobile phone images.

In this dataset there are 3500 plus images of each currency.

# 4. Dataset Preparation and Preprocessing

The images are prepared in YOLO object detection format. Each annotated object contains a class identifier and normalized bounding box coordinates.

Typical preprocessing and augmentation include:

1. Resizing images to the model input resolution.
2. Normalizing image data according to the model implementation.
3. Brightness and contrast variations.
4. Small rotation and geometric variations.
5. Scaling and cropping variations.
6. Updating bounding boxes correctly after augmentation.

Lighting is particularly important because glare can make security threads and other printed features difficult to detect.


# 5. Dataset Challenges

The main challenges identified during development are:

- Unequal numbers of annotations between classes.
- Small security features compared with the complete banknote.
- Glare and excessive brightness.
- Blurred mobile camera images.
- Different camera angles.
- Poor illumination.
- Simulated counterfeit images may not represent every characteristic of real counterfeit notes.

A missing model detection therefore should not automatically be interpreted as proof that a banknote is counterfeit. The feature may be hidden because of image quality.


# 6. YOLO11s Model Description

The primary object detection model used in the current application is **YOLO11s**.

YOLO11s is a lightweight member of the YOLO object detection family. It is fine tuned on the Pakistani currency dataset so that it can detect banknote denominations and security features.

The detector provides:

- Detected class
- Confidence score
- Bounding box coordinates

The model is trained to detect denomination, SBP seals, serial numbers, printed year, security thread, security lines and security dots.


# 7. Why YOLO11s Is Used

The application needs to identify the location of several features within a single banknote image. Object detection is suitable because it provides both the feature identity and its location.

For example, the system needs to locate the SBP seal, serial number and security thread rather than only classify the entire image.

The small YOLO11s model provides a practical balance between detection capability, inference speed and computational requirements for an FYP mobile prototype.


# 10. Rule Based Feature Verification

After YOLO detection, the application applies denomination dependent verification rules.

Every scanned side is checked for the required general features:

- SBP seal
- Serial number
- Printed year

The current project prototype uses these additional feature rules:

| Denomination | Additional security features |
|---|---|
| Rs. 10 | None of thread, lines or dots |
| Rs. 20 | Security lines |
| Rs. 50 | Security lines |
| Rs. 100 | Security thread and security lines |
| Rs. 500 | Security thread and security dots |
| Rs. 1000 | Security thread and security dots |
| Rs. 5000 | Security thread and security dots |

These rules should be checked against the exact banknote series used in the project before being presented as official characteristics.


# 13. Evaluation Metrics

For YOLO11s, the main evaluation metrics are:

- Precision
- Recall
- mAP@50
- mAP@50-95

**Precision** measures how many predicted objects are correct.

**Recall** measures how many relevant objects are detected.

**mAP@50** measures mean average precision at an IoU threshold of 0.50.

**mAP@50-95** averages performance across IoU thresholds from 0.50 to 0.95.

For MobileNetV3, suitable metrics include accuracy, precision, recall, F1-score and a confusion matrix.


# 16. Limitations

This project is an academic prototype. Computer vision predictions can be affected by:

- Lighting
- Camera quality
- Focus
- Image angle
- Occlusion
- Damaged banknotes
- Differences between training and real-world images

The application should therefore report **feature verification** rather than claim an absolute guarantee of authenticity.
